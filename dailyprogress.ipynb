```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/Daily_progress_chart.csv")
```

```python
df.head()
print (df)
```

```python
df.columns
```

```python
# manipulation and preprocessing dataframe
df["Total_time_quran_ED"] = df ["namaz_unit_ED"] * df ["Unit_time_namaz_ED"]
df["W_namaz_ED"] =  df ["Total_time_namaz_ED"] / df ["Total_time"] * 10 
df["W_namaz_ED"] = df["W_namaz_ED"].round(2) 

df["Total_time_namaz_ED"] = df ["quran_unit_ED"] * df ["Unit_time_quran_ED"]
df["W_quran_ED"] =  df ["Total_time_namaz_ED"] / df ["Total_time"] * 10 
df["W_quran_ED"] = df["W_quran_ED"].round(2) 

df["Total_time_eating_ED"] = df ["eating_unit_ED"] * df ["Unit_time_eating_ED"]
df["W_eating_ED"] =  df ["Total_time_eating_ED"] / df ["Total_time"] * 10 
df["W_eating_ED"] = df["W_eating_ED"].round(2) 

df["Total_time_sleep_ED"] = df ["sleep_unit_ED"] * df ["Unit_time_sleep_ED"]
df["W_sleep_ED"] =  df ["Total_time_sleep_ED"] / df ["Total_time"] * 10 
df["W_sleep_ED"] = df["W_sleep_ED"].round(2)

df["Total_time_exercise_ED"] = df ["exercise_unit_ED"] * df ["Unit_time_exercise_ED"]
df["W_exercise_ED"] =  df ["Total_time_eating_ED"] / df ["Total_time"] * 10 
df["W_exercise_ED"] = df["W_exercise_ED"].round(2)

df["Total_time_office_work_ED"] = df ["office_work_unit_ED"] * df ["Unit_time_office_work_ED"]
df["W_office_work_ED"] =  df ["Total_time_office_work_ED"] / df ["Total_time"] * 10 
df["W_office_work_ED"] = df["W_office_work_ED"].round(2)

df["Total_time_coding_ED"] = df ["coding_unit_ED"] * df ["Unit_time_coding_ED"]
df["W_coding_ED"] =  df ["Total_time_coding_ED"] / df ["Total_time"] * 10 
df["W_coding_ED"] = df["W_coding_ED"].round(2)

df["Total_time_sb_ED"] = df ["study_book_unit_ED"] * df ["Unit_time_sb_ED"]
df["W_sb_ED"] =  df ["Total_time_sb_ED"] / df ["Total_time"] * 10
df["W_sb_ED"] = df["W_sb_ED"].round(2)

df["Total_time_WS_ED"] = df ["Writing_Speaking_unit_ED"] * df ["Unit_time_WS_ED"]
df["W_writing_speakig_ED"] =  df ["Total_time_WS_ED"] / df ["Total_time"] * 10
df["W_writing_speakig_ED"] = df["W_writing_speakig_ED"].round(2)

df["family_total_time_ED"] = df ["Family_unit_time_ED"] * df ["family__time_ED"]
df["W_family_time_ED"] =  df ["family_total_time_ED"] / df ["Total_time"] * 10
df["W_family_time_ED"] = df["W_family_time_ED"].round(2)

df["Sum_of_OD"] = df ["W_namaz_OD"] + df ["W_quran_OD"]+ df ["W_eating_OD"] + df ["W_sleeping_OD"] + df ["W_exercise_OD"] + df ["W_coffice_work_OD"] + df ["W_coding_OD"] + df ["W_study_book_OD"] + df ["W_writing_speakig_OD"] + df ["W_family_time_OD"]
df["Sum_of_OD"] = df["Sum_of_OD"].round(2)


df["Sum_of_ED"] = df ["W_namaz_ED"] + df ["W_quran_ED"]+ df ["W_eating_ED"] + df ["W_sleep_ED"] + df ["W_exercise_ED"] + df ["W_office_work_ED"] + df ["W_coding_ED"] + df ["W_sb_ED"] + df ["W_writing_speakig_ED"] + df ["W_family_time_ED"]
df["Sum_of_ED"] = df["Sum_of_OD"].round(2)
```
